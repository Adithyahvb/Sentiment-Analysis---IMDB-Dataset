# Sentiment Analysis Model Performance Comparison

Three deep learning models were developed and evaluated for IMDb movie review sentiment classification: a Custom LSTM model, a ULMFiT model, and a BERT-based transformer model. The performance of each model was measured using Accuracy, Precision, Recall, and F1-Score on the test dataset containing 25,000 reviews.

---

## 1. Custom LSTM Model Performance

The Custom LSTM model achieved a test accuracy of **79.80%**. Both the Positive and Negative sentiment classes obtained a precision, recall, and F1-score of **0.80**, indicating balanced classification performance across both categories.

### Observations

* The model demonstrated stable and consistent predictions.
* Since LSTM models process text sequentially, they are effective in capturing contextual dependencies in reviews.
* However, the model struggled with more complex contextual relationships and long-range semantic understanding compared to transformer-based architectures.

### Metrics

* **Accuracy:** 79.80%
* **Precision:** 0.80
* **Recall:** 0.80
* **F1-Score:** 0.80

---

## 2. ULMFiT Model Performance

The ULMFiT model significantly improved performance over the Custom LSTM model, achieving a test accuracy of **85.34%**.

For the Negative class, the model obtained:

* Precision: 0.88
* Recall: 0.82
* F1-score: 0.85

For the Positive class:

* Precision: 0.83
* Recall: 0.89
* F1-score: 0.86

### Observations

* ULMFiT benefited from transfer learning using a pretrained language model.
* The model showed better contextual understanding and generalization than the Custom LSTM model.
* Higher recall for the Positive class indicates the model effectively identified positive sentiments with fewer false negatives.
* The model achieved a good balance between precision and recall.

### Metrics

* **Accuracy:** 85.34%
* **Precision:** 0.86
* **Recall:** 0.85
* **F1-Score:** 0.85

---

## 3. BERT Model Performance

The BERT model achieved the best overall performance among all three models with a test accuracy of **92.28%**.

### Metrics

* **Accuracy:** 92.28%
* **Precision:** 0.9082
* **Recall:** 0.9406
* **F1-Score:** 0.9241

### Observations

* BERT outperformed both LSTM and ULMFiT models due to its bidirectional transformer architecture.
* The model captured deep semantic and contextual relationships in movie reviews more effectively.
* High recall indicates that the model successfully identified most positive sentiment instances.
* The strong F1-score demonstrates an excellent balance between precision and recall.

---
# Overall Comparison

The experimental results clearly show progressive improvements from the Custom LSTM model to ULMFiT and finally to BERT.

* The **Custom LSTM** model provided a solid baseline performance but was limited in understanding complex contextual semantics.
* **ULMFiT** improved classification accuracy through transfer learning and pretrained language representations.
* **BERT** achieved the highest performance due to its transformer-based bidirectional contextual understanding.

BERT improved accuracy by approximately:

* **12.48%** over the Custom LSTM model.
* **6.94%** over the ULMFiT model.

These results demonstrate that transformer-based architectures are highly effective for sentiment analysis tasks, particularly when dealing with large-scale textual datasets such as IMDb movie reviews.

---

# Conclusion

Among the three evaluated models, the BERT model produced the best sentiment classification performance across all evaluation metrics. Its superior contextual understanding, semantic representation capability, and transformer-based architecture enabled highly accurate sentiment predictions.

Although the Custom LSTM and ULMFiT models performed reasonably well, BERT proved to be the most reliable and robust solution for IMDb sentiment analysis.
